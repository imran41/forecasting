```python
Walmart["months"] = 0

for i in range(len(Walmart["Footfalls"])):
  p = Walmart["Month"][i]
  Walmart["months"][i] = p[0:3]

Walmart.head()
```

```python
Walmart["t"] = np.arange(1, len(Walmart["Footfalls"]) + 1)
Walmart["t_square"] = Walmart["t"] ** 2
Walmart["log_Footfalls"] = np.log(Walmart["Footfalls"])


# Sin = np.sin((2 * pi * Walmart["t"]) / 12)
# Cos = np.cos((2 * pi * Walmart["t"]) / 12)

# Walmart["Sin"] = Sin
# Walmart["Cos"] = Cos


Walmart
```

```python
month_dummies = pd.DataFrame(pd.get_dummies(Walmart["months"]))
Walmart1 = pd.concat([Walmart, month_dummies], axis = 1)
Walmart1
```

```python
Train = Walmart1[:147]
Train
```

```python
Test = Walmart1[147:]
Test
```

```python
import statsmodels.formula.api as smf
```

```python
linear_model = smf.ols("Footfalls ~ t", data = Train).fit()
print(linear_model.summary())
```

### Linear

```python
pred = linear_model.predict(Test["t"])
pred
```

```python
rmse_linear = root_mean_squared_error(Test["Footfalls"], pred)
rmse_linear
```

### Expo

```python
exp_model = smf.ols("log_Footfalls ~ t", data = Train).fit()
print(exp_model.summary())
```

```python
pred_exp = exp_model.predict(Test["t"])
rmse_Exp = root_mean_squared_error(Test["Footfalls"], np.exp(pred_exp))
rmse_Exp
```

#### Quadratic

```python
quad_model = smf.ols("Footfalls ~ t + t_square", data = Train).fit()
print(quad_model.summary())
```

```python
pred_quad_model = quad_model.predict(Test[["t", "t_square"]])
rmse_Quad = root_mean_squared_error(Test["Footfalls"], pred_quad_model)
rmse_Quad
```

### Additive seasonality

```python
add_sea = smf.ols('Footfalls ~ Jan+Feb+Mar+Apr+May+Jun+Jul+Aug+Sep+Oct+Nov', data=Train).fit()
print(add_sea.summary())
```

```python
pred_add_sea = add_sea.predict(Test[['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov']])
rmse_add_sea = root_mean_squared_error(Test['Footfalls'], pred_add_sea)
rmse_add_sea
```

### Multiplicative Seasonality

```python
Mul_sea = smf.ols('log_Footfalls ~ Jan+Feb+Mar+Apr+May+Jun+Jul+Aug+Sep+Oct+Nov',data = Train).fit()
print(Mul_sea.summary())
```

```python
pred_Mult_sea = Mul_sea.predict(Test)
rmse_Mult_sea = root_mean_squared_error(Test['Footfalls'], np.exp(pred_Mult_sea))
rmse_Mult_sea
```

### Additive Seasonality Quadratic Trend

```python
add_sea_Quad = smf.ols('Footfalls ~ t+t_square+Jan+Feb+Mar+Apr+May+Jun+Jul+Aug+Sep+Oct+Nov', data=Train).fit()
print(add_sea_Quad.summary())
```

```python
pred_add_sea_quad = pd.Series(add_sea_Quad.predict(Test[['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','t','t_square']]))
rmse_add_sea_quad = root_mean_squared_error(Test['Footfalls'], pred_add_sea_quad)
rmse_add_sea_quad
```

```python
# Autoregression Model (AR)
# Calculating Residuals from best model applied on full data
# AV - FV
full_res = Walmart1.Footfalls - model_full.predict(Walmart1)
full_res
```

```python
import matplotlib.pyplot as plt
import statsmodels.graphics.tsaplots as tsa_plots

# Create a subplot with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ACF plot on left side
tsa_plots.plot_acf(full_res, lags=12, ax=axes[0])
axes[0].set_title("ACF (Autocorrelation Function)")

# PACF plot on right side
tsa_plots.plot_pacf(full_res, lags=12, ax=axes[1])
axes[1].set_title("PACF (Partial Autocorrelation Function)")

# Adjust layout
plt.tight_layout()
plt.show()
```

Yeh dono graphs ACF aur PACF ka difference visually dikhate hain 👇

---

### 🔹 **Left side: ACF (Autocorrelation Function)**

* Bars gradually **decrease** jaise lag 1, 2, 3... badhta hai.
* Matlab current value apne **previous values se indirectly bhi related hai**.
* Example: Agar kal sale zyada tha → aaj bhi zyada → parso bhi thoda effect.

---

### 🔹 **Right side: PACF (Partial Autocorrelation Function)**

* Sirf **lag 1** bar strong hai, baaki zero ke paas.
* Matlab current sale **sirf directly 1 day pehle ke sale se related** hai,
  aur baaki lags ka effect indirectly aata hai.

---

### 🔹 **Easy Hinglish Summary:**

* **ACF:** total relationship with all previous days (direct + indirect).
* **PACF:** sirf direct relationship with that lag day.

---

Ye pattern AR(1) process ka example hai —
isliye PACF me sirf **lag 1 significant** hai,
aur ACF gradually decline karta hai.


```python
from statsmodels.tsa.ar_model import AutoReg
```

```python
model_ar = AutoReg(full_res, lags = [1])
model_fit = model_ar.fit()

print("Coefficients:\n%s" % model_fit.params)
```

```python
pred_res = model_fit.predict(start = len(full_res),
                             end = len(full_res) + len(predict_data) - 1,
                             dynamic = False)

pred_res.reset_index(drop = True, inplace = True)
pred_res
```

```python
final_pred = pred_new + pred_res
final_pred
```
