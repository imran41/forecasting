```python
#a = pd.DataFrame(data.groupby(data["week"]).median())
a = pd.DataFrame(data.groupby(data["week"]).mean())
a.reset_index(inplace = True)
df3 = a[["week", "Sales volume in Tonnes"]]
```

```python
df3["week"] = df3["week"].astype(int)
df3.sort_values(ascending = True, by = "week", inplace = True)
```

```python
df3.rename(columns = {"Sales volume in Tonnes": "Sales_volume_in_Tonnes_MEAN"}, inplace = True)
df3.reset_index(inplace = True)
```

```python
df3.drop(columns = "index", inplace = True)
df3
```

```python
plt.figure(figsize = (9,3))
sns.lineplot(x = df3["week"], y = df3["Sales_volume_in_Tonnes_MEAN"])

plt.tight_layout()
plt.show()
```

```python
df3["t"] = np.arange(1, len(df3) + 1)
df3["t2"] = df3["t"]**2
df3["log_value"] = np.log(df3["Sales_volume_in_Tonnes_MEAN"])
df3
```

```python
from cmath import pi

Sin = np.sin((2 * pi * df3["t"]) / 52.143)
Cos = np.cos((2 * pi * df3["t"]) / 52.143)

df3["Sin"] = Sin
df3["Cos"] = Cos

df3
```

Linear Model

```python
model = smf.ols("Sales_volume_in_Tonnes_MEAN ~ t", data = df3).fit()
print(model.summary())
```

```python
print("For Test Data")
pred = model.predict(df3)
print("RMSE: ", np.sqrt(mean_squared_error(df3["Sales_volume_in_Tonnes_MEAN"], pred)))
print("MAPE: ", mean_absolute_percentage_error(df3["Sales_volume_in_Tonnes_MEAN"], pred) * 100)
```

Exponential

```python
model = smf.ols("log_value ~ t", data = df3).fit()
print(model.summary())
```

```python
print("For Test Data")
pred = model.predict(df3)
print("RMSE: ", np.sqrt(mean_squared_error(df3["Sales_volume_in_Tonnes_MEAN"], np.exp(pred))))
print("MAPE: ", mean_absolute_percentage_error(df3["Sales_volume_in_Tonnes_MEAN"], np.exp(pred)) * 100)
```

Quadratic

```python
model = smf.ols("Sales_volume_in_Tonnes_MEAN ~ t + t2", data = df3).fit()
print(model.summary())
```

```python
print("For Test Data")
pred = model.predict(df3)
print("RMSE: ", np.sqrt(mean_squared_error(df3["Sales_volume_in_Tonnes_MEAN"], pred)))
print("MAPE: ", mean_absolute_percentage_error(df3["Sales_volume_in_Tonnes_MEAN"], pred) * 100)
```
